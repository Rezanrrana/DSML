```python
data1 = ['ram', 'shyam', 'gita', 'rita']
data2 = [20,30,40,50]

for i in zip(data1, data2):
  print(i)
```

```python
for i, (name,age) in enumerate(zip(data1,data2), start=1):
  print(i,name,age)
```

```python
# classwork
# find the maximum number using reduce fucntion

lst = [4,9,2,8,1]

def maximum(x,y):
  if x > y:
    return x
  return y
```

```python
# (2D) area = l * w
#  (3D) area l *w*h
# solve using the position argument
```

```python
def area(*args):
  if len(args) ==2:
    l,w = args
    print("2D Area = ", l*w)
  elif len(args) == 3:
    l,w,h = args
    print("3D volume = ", l*w*h)
  else:
    print("Invalid number of argument")
```

```python
area(4,5)
area(4,5,6)
```

```python
area(2)
```

```python
# multiply


def mux(*args):
  result = 1
  for num in args:
    result *= num
  return result

mux(2,3,4,5)
```

```python
mux(1,2,3)
```

```python
mux(1,2,3,4,5,6,7,8,9)
```
