Student Grade Management System

Create a Python class named Student.

Requirements:

1. Create a class attribute named school.

2. Use the init() method to take:
*   Student name
*   Student marks

3. Create an instance method named check_grade() that displays:
>  A Grade: Marks ≥ 80

>  B Grade: Marks ≥ 60 and < 80

>  C Grade: Marks ≥ 40 and < 60

>  Fail: Marks < 40
4. Create a class method to display the school name.
5. Create a static method to display a welcome message.
6. Create an object and display the student's grade.


```python
class Student:
    school = "Tribhuwan High School"                     # 1. class attribute

    def __init__(self, name, marks):                        # 2. constructor
        self.name = name
        self.marks = marks

    def check_grade(self):                                  # 3. instance method
        if self.marks >= 80:
            print(f"{self.name}'s Grade: A")
        elif self.marks >= 60:
            print(f"{self.name}'s Grade: B")
        elif self.marks >= 40:
            print(f"{self.name}'s Grade: C")
        else:
            print(f"{self.name}'s Grade: Fail")

    @classmethod
    def show_school(cls):                                    # 4. class method
        print("School Name:", cls.school)

    @staticmethod
    def welcome():                                            # 5. static method
        print("Welcome to the Student Grade Management System \n")


# 6. Create an object and display the student's grade
Student.welcome()          # static method call
Student.show_school()      # class method call

student1 = Student("Anisha", 85)
student1.check_grade()

student2 = Student("Rohan", 55)
student2.check_grade()
```
