```python
# # factorial number

# 5! = 5*4*3*2*1 =120
# 7! =7*6*5!

# 1) n* (n-1) (recursive case)
# 2)  1 or 0 (base case)
```

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)

factorial(5)
```

```python
def recur_fibo(n):
  if n<=1:
    return n
  else:
    return(recur_fibo(n-1)+recur_fibo(n-2))
```

```python
nterms = int(input("Enter a number"))

if nterms <=0:
  print("please enter a postive integer")
else:
  print("Fibonacci sequence:")
  for i in range(nterms):
    print(recur_fibo(i))
```

```python
def myWrapper(func):
  def myInnerFunc():
    print("Inner wrapper.")
    func()
  return myInnerFunc

@myWrapper
def myFunc():
  print("Hello, world!")

myFunc()
```

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"Calling {func.__name__}")
    return func(*args, **kwargs)
  return wrapper

@logger
def greet(name,age):
  print(f"Hello {name}, Age: {age},")

greet("venzi_X", 23)
```
