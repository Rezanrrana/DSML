```python
def sum(*args):
  result = 0
  for num in args:
    result += num
  return result
```

```python
def introduce(**kwargs):
  detail = []
  for k,v in kwargs.items():
    print(k,v)
    detail.append(k + ":" + str(v))
  return ",". join(detail)
```

```python
introduce(Name='Raju', Age= '25', city = 'Ktm')
```

**Write a function that accepts products and their keyword arguments and returns the total**

```python
def product(**kwargs):
  total = 0
  for values in kwargs.values():
    total += values
  return total
```

```python
result = product(rice =500, meat =1300, Floar = 400)
print("total:", result)
```
